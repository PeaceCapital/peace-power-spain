```python
import requests, time
import pandas as pd
import numpy as np
from io import StringIO

def fetch_omie_day(date):
    d = date.strftime("%d_%m_%Y")
    url = (
        f"https://www.omie.es/sites/default/files/dados/"
        f"AGNO_{date.year}/MES_{date.month:02d}/TXT/"
        f"INT_PBC_EV_H_1_{d}_{d}.TXT"
    )
    for attempt in range(3):
        try:
            r = requests.get(url, timeout=15)
            if r.status_code == 200:
                return r.text
            time.sleep(1)
        except:
            time.sleep(2)
    return None

records = []
date = pd.Timestamp("2023-01-01")
end  = pd.Timestamp("2026-04-09")
errors = 0

while date <= end:
    raw = fetch_omie_day(date)
    if raw:
        for line in raw.strip().split("\n")[1:]:
            parts = line.strip().split(";")
            if len(parts) >= 4 and parts[0].strip() == "1":
                try:
                    hour  = int(parts[2].strip()) - 1
                    price = float(parts[3].strip().replace(",", "."))
                    records.append({
                        "datetime": date + pd.Timedelta(hours=hour),
                        "price_omie": price
                    })
                except:
                    pass
    else:
        errors += 1
    
    if len(records) % 2400 == 0 and records:
        print(f"  {date.date()} — {len(records):,} hours collected")
    
    date += pd.Timedelta(days=1)
    time.sleep(0.15)  # polite delay

print(f"\n✓ Done — {len(records):,} hours, {errors} errors")

spot = (
    pd.DataFrame(records)
    .set_index("datetime")["price_omie"]
    .sort_index()
    .astype(float)
)
spot.index = pd.DatetimeIndex(spot.index).tz_localize(
    "Europe/Madrid", ambiguous="NaT", nonexistent="shift_forward"
)
spot = spot[spot.index.notna()]
print(f"✓ Hours: {len(spot):,}")
print(f"✓ Range: {spot.index[0]} → {spot.index[-1]}")
```
