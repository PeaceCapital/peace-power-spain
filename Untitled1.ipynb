```python
import requests

import pandas as pd

TOKEN = 'f6835be28cb33be5b677a4f8866bec14955f81f7ff39cac08880a8c2fc474792'

HEADERS = {

    'x-api-key': TOKEN,

    'Accept': 'application/json',

}

INDICATORS = {

    600:   'omie_price',        # OMIE Spain DA price

    541:   'gen_wind',          # Wind generation peninsular

    542:   'gen_solar_pv',      # Solar PV generation

    10034: 'gen_solar_thermal', # Solar thermal generation

    138:   'gen_hydro',         # Hydro generation

    474:   'gen_nuclear',       # Nuclear generation

    2200:  'demand_forecast',   # Forecast demand

    10209: 'atc_es_fr',         # ATC ES->FR interconnector

}

START = '2025-05-01T00:00:00+02:00'

END   = '2025-05-07T23:00:00+02:00'

results = {}

for ind_id, name in INDICATORS.items():

    r = requests.get(

        'https://api.esios.ree.es/indicators/{}'.format(ind_id),

        headers=HEADERS,

        params={'start_date': START, 'end_date': END, 'time_trunc': 'hour'},

        timeout=15,

    )

    if r.status_code == 200:

        values = r.json()['indicator']['values']

        df = pd.DataFrame(values)[['datetime', 'value']]

        df.columns = ['datetime', name]

        df['datetime'] = pd.to_datetime(df['datetime'])

        df = df.set_index('datetime')

        results[name] = df

        print('OK {:>5} — {:>22} — {:>4} rows — last value: {:.2f}'.format(

            ind_id, name, len(df), float(df[name].iloc[-1])

        ))

    else:

        print('FAIL {:>5} — {:>22} — status {}'.format(ind_id, name, r.status_code))

# Combine into one DataFrame

live = pd.concat(results.values(), axis=1)

print()

print('Live DataFrame: {} rows x {} cols'.format(len(live), len(live.columns)))

print(live.tail(3))
```

```python
# Merge on datetime index, outer join, resample to hourly
dfs = list(results.values())

live = dfs[0]
for df in dfs[1:]:
    live = live.merge(df, left_index=True, right_index=True, how='outer')

# Resample to clean hourly index
live = live.resample('h').mean()

# Forward fill gaps (max 3 hours)
live = live.ffill(limit=3)

print('Live DataFrame: {} rows x {} cols'.format(len(live), len(live.columns)))
print()
print(live.tail(5).to_string())
```

```python
# ── Final indicator set ───────────────────────────────────────
LIVE_INDICATORS = {
    600:   ('omie_price',         {'geo_ids[]': 3}),   # Spain only
    541:   ('gen_wind',           {}),
    542:   ('gen_solar_pv',       {}),
    10034: ('gen_solar_thermal',  {}),
    138:   ('gen_hydro',          {}),
    474:   ('gen_nuclear',        {}),
    544:   ('demand_forecast',    {}),
    10209: ('atc_es_fr',          {}),
}

START = '2025-05-01T00:00:00+02:00'
END   = '2025-05-07T23:00:00+02:00'

frames = {}

for ind_id, (name, extra_params) in LIVE_INDICATORS.items():
    params = {
        'start_date': START,
        'end_date':   END,
        'time_trunc': 'hour',
        **extra_params,
    }
    r = requests.get(
        'https://api.esios.ree.es/indicators/{}'.format(ind_id),
        headers=HEADERS,
        params=params,
        timeout=15,
    )
    if r.status_code != 200:
        print('FAIL {:>5} — {} — status {}'.format(ind_id, name, r.status_code))
        continue

    values = r.json()['indicator']['values']
    df = pd.DataFrame(values)[['datetime', 'value']]
    df.columns = ['datetime', name]
    df['datetime'] = pd.to_datetime(df['datetime'])
    df = df.set_index('datetime')
    df = df[~df.index.duplicated(keep='last')]
    frames[name] = df
    print('OK {:>5} — {:>22} — {:>4} rows — last: {:.2f}'.format(
        ind_id, name, len(df), float(df[name].iloc[-1])
    ))

# ── Merge to hourly ───────────────────────────────────────────
live = list(frames.values())[0]
for df in list(frames.values())[1:]:
    live = live.merge(df, left_index=True, right_index=True, how='outer')

live = live.resample('h').mean()
live = live.ffill(limit=3)

# ── Derived features ──────────────────────────────────────────
# Generation in MWh/h = MW effective
live['gen_renewable']   = (live['gen_solar_pv']
                           + live['gen_solar_thermal']
                           + live['gen_wind'])
live['rsi']             = live['gen_renewable'] / live['demand_forecast'].clip(lower=1)
live['thermal_floor']   = 35.0 * 0.45 + 65.0 * 0.35   # TTF proxy + EUA proxy
live['floor_discount']  = live['omie_price'] - live['thermal_floor']
live['atc_util']        = (live['atc_es_fr'].abs()
                           / live['atc_es_fr'].abs().rolling(168).max().clip(lower=1))

# Regime
def regime(row):
    if row['rsi'] > 0.65:
        return 'Renewable-Dom.'
    elif row['omie_price'] > row['thermal_floor'] * 1.3:
        return 'Demand-Stress'
    else:
        return 'Thermal-Marginal'

live['regime'] = live.apply(regime, axis=1)

print()
print('Live DataFrame: {} rows x {} cols'.format(len(live), len(live.columns)))
print()
print(live[['omie_price','gen_renewable','rsi',
            'thermal_floor','floor_discount','regime']].tail(5).to_string())
```
