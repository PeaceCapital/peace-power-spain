# Iberia Data Pull Quickstart

Use this notebook to pull OMIE prices and search or fetch ESIOS indicators without going through the websites manually.

```python
from peace_power import fetch_omie_marginal_prices, search_esios_indicators, fetch_esios_indicator

ESIOS_TOKEN = "YOUR_ESIOS_TOKEN"
```

```python
omie = fetch_omie_marginal_prices(
    "2026-03-01",
    "2026-03-07",
    output_dir="../data/raw/omie",
)
omie.head()
```

```python
matches = search_esios_indicators("bateria", token=ESIOS_TOKEN)
matches[["id", "name", "short_name"]].head(10)
```

```python
indicator_id = int(matches.iloc[0]["id"])

bess_online = fetch_esios_indicator(
    indicator_id=indicator_id,
    token=ESIOS_TOKEN,
    start_date="2026-03-01T00:00:00+01:00",
    end_date="2026-03-07T23:00:00+01:00",
    time_trunc="hour",
)
bess_online.head()
```
